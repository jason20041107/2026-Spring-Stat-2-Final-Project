## Data Selection

For this project, only the variables related to the research question were selected from the original YRBS 2007 dataset.

The research topic is the relationship between physical activity and BMI percentile among high school students. Therefore, the selected variables include physical activity, BMI percentile, sex, and grade level.

| Variable | Description | Purpose |
|---|---|---|
| PhysicalActivity5OrMoreDays | Number of days students were physically active for at least 60 minutes during the past 7 days | Main independent variable |
| BMIPCT | BMI percentile by age and sex | Dependent variable |
| WhatIsYourSex | Student's sex | Control variable |
| InWhatGradeAreYou | Student's grade level | Control variable |

These variables were selected because they are directly related to the research question and can be used for regression analysis.

```python
import pandas as pd
import numpy as np
import os

# 原始資料路徑
raw_path = r"C:\Users\curry\OneDrive\桌面\統計實習\2026-Spring-Stat-2-Final-Project\data\data raw\YRBS_2007.csv"

# 輸出資料夾路徑
processed_dir = r"C:\Users\curry\OneDrive\桌面\統計實習\2026-Spring-Stat-2-Final-Project\data\data processed"

# 建立輸出資料夾
os.makedirs(processed_dir, exist_ok=True)

# 讀取資料
df = pd.read_csv(raw_path)

# 看前五筆資料
df.head()
```

## Data Cleaning and Recoding

After selecting the required variables, the dataset was cleaned and recoded to make it suitable for statistical analysis.

First, missing values were removed to avoid errors in the regression model. Then, the physical activity variable was recoded from the original survey code into the actual number of days.

In the original dataset, the physical activity variable uses codes from 1 to 8. These codes represent 0 to 7 days of physical activity. Therefore, the new variable `PhysicalActivityDays` was created by subtracting 1 from the original code.

Sex and grade variables were also recoded into readable labels to make the results easier to understand. Finally, the BMI percentile variable was renamed as `BMIPercentile`.

The cleaned dataset includes the following final variables:

| New Variable | Description |
|---|---|
| PhysicalActivityDays | Actual number of physically active days, from 0 to 7 |
| BMIPercentile | BMI percentile by age and sex |
| Sex | Student's sex |
| Grade | Student's grade level |

```python
# 選擇這次研究主題需要的欄位
cols = [
    "PhysicalActivity5OrMoreDays",
    "BMIPCT",
    "WhatIsYourSex",
    "InWhatGradeAreYou"
]

data = df[cols].copy()

# 移除缺失值
data = data.dropna()

# 將運動天數重新編碼
# 原始代碼 1~8 對應真正天數 0~7
data["PhysicalActivityDays"] = data["PhysicalActivity5OrMoreDays"] - 1

# 重新編碼性別
data["Sex"] = data["WhatIsYourSex"].map({
    1: "Female",
    2: "Male"
})

# 重新編碼年級
data["Grade"] = data["InWhatGradeAreYou"].map({
    1: "9th grade",
    2: "10th grade",
    3: "11th grade",
    4: "12th grade",
    5: "Other"
})

# 重新命名 BMI 欄位
data = data.rename(columns={
    "BMIPCT": "BMIPercentile"
})

# 保留整理後欄位
cleaned_data = data[
    ["PhysicalActivityDays", "BMIPercentile", "Sex", "Grade"]
]

cleaned_data.head()
```

## Saving the Cleaned Dataset

After data selection, cleaning, and recoding, the cleaned dataset was saved as a new CSV file.

This processed dataset will be used for descriptive statistics, data visualization, and regression analysis in the next steps.

```python
output_path = os.path.join(processed_dir, "cleaned_yrbs_physical_activity_bmi.csv")

cleaned_data.to_csv(output_path, index=False, encoding="utf-8-sig")

print("Cleaned data saved to:")
print(output_path)
```
